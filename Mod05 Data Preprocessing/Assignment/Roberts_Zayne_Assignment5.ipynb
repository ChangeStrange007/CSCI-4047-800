```python
import pandas as pd, numpy, sklearn
from sklearn import preprocessing
from sklearn.impute import KNNImputer

df = pd.read_csv('Income Dirty Data.csv')
df.head()
```

3. Calculate and display the following information
    - Total number of NaN values for **each column**
    - Percentage of NaN values in the dataset 
    - Number of rows *without* any NaN values

```python
def missing_values():
    print("Missing values per column:")
    null_values = df.isna().sum()
    print(null_values.to_string())

    print("\nPercentage of missing values:")
    pct_missing = null_values.sum()/(len(df) * len(df.columns))
    print(f"{pct_missing * 100:.2F}%")

    print("\nNumber of rows WITHOUT any missing values:")
    rows_with_missing = len((df[(df.income.isna()) | (df.tax_15_pct.isna())]))
    print(len(df) - rows_with_missing)

missing_values()
```

Besides missing values (NaN), the dataset contains errors. We have the following rules to check:

* All employees are adults (18+ years old)
* All employees pay 15% of their income for the tax
* All employees make money; no income should be <= 0 
---
4. Calculate and display the percentage of the data that does **NOT** violate any **one** of the rules.

```python
def check_rules():
    age_ok = df['age'] >= 18
    income_ok = df['income'] > 0
    tax_ok = numpy.isclose(df['tax_15_pct'], df['income'] * 0.15)

    valid = age_ok & income_ok & tax_ok
    pct_valid = valid.sum() / len(df) * 100

    print(f"Percentage of rows that do NOT violate any rule: {pct_valid:.2f}%")

check_rules()
```

Now that we have determined the number of erroneous datapoints in our set, let's work on correcting it as best we can.

5. Replace non *Female*/*Male* values in the **Sex** column with either *Female* or *Male* (e.g., Women --> Female)

```python
sex_map = {
    'Women': 'Female',
    'Woman': 'Female',
    'Men': 'Male',
    'Man': 'Male'
}

df['sex'] = df['sex'].replace(sex_map)
print(df['sex'].value_counts(dropna=False))
```

6. Replace non-positive **Age** values with NaN (`numpy.NaN`)
7. Replace non-positive **Income** values with NaN (`numpy.NaN`)
8. Replace non-positive **Tax (15%)** values with NaN (`numpy.NaN`)

```python
df.loc[df['age'] <= 0, 'age'] = numpy.nan
df.loc[df['income'] <= 0, 'income'] = numpy.nan
df.loc[df['tax_15_pct'] <= 0, 'tax_15_pct'] = numpy.nan

missing_values()
```

9. Use machine learning (`KNNImputer`) to impute all missing values (replaces NaN values with the most predicted values)
    - Will need to use a scaler and convert the values in the **Sex** column to a numeric value for algorithm to work properly
    - Show some of the data prior to imputing, and after imputing

```python
print("Before imputing:")
display(df.head(10))

df_impute = df.copy()
df_impute['sex'] = df_impute['sex'].map({'Female': 0, 'Male': 1})

feature_cols = ['sex', 'age', 'income', 'tax_15_pct']

scaler = preprocessing.StandardScaler()
scaled = scaler.fit_transform(df_impute[feature_cols])

imputer = KNNImputer(n_neighbors=5)
imputed_scaled = imputer.fit_transform(scaled)

imputed = scaler.inverse_transform(imputed_scaled)

df_imputed = df_impute.copy()
df_imputed[feature_cols] = imputed

df_imputed['sex'] = df_imputed['sex'].round().map({0: 'Female', 1: 'Male'})
df_imputed['age'] = df_imputed['age'].round().astype(int)
df_imputed['income'] = df_imputed['income'].round(2)
df_imputed['tax_15_pct'] = df_imputed['tax_15_pct'].round(2)

print("\nAfter imputing:")
display(df_imputed.head(10))
```

10. In the empty `Markdown` cell below, explain why it is important to clean a dataset before calculating analytics about the data

Cleaning a dataset before calculating analytics is important because missing values, inconsistent categories, and impossible values distort summary statistics like means, percentages, and correlations. If these errors are not corrected first, any analysis using the data will be wrong and decisions made from that analysis could be incorrect. Cleaning the data first ensures that the patterns we find actually reflect the real population rather than data mistakes or gaps in the collection.
