To measure the psychological state of Americans, a psychology researcher (in 1929) devised
three questions put to 3443 Americans from different walks of life. He asked them whether they
were happy. They could answer only **yes** or **no** to each question. By today’s standards the study
is quite simple, and he did no formal statistical analysis of his data. The frequency data (number
of **yes** and **no** responses within each employment category) from this study are in the file
*Americandata.csv*.

---

You need to use the data and answer the question, that is, **Is the type of profession related to the happiness of American people?**
Report the **χ2** statistic and the *statistical significance of test*, in addition to the contingency table.
Also, explain the results and what they mean. What does the value of **χ2** statistic mean? Is it
good or bad? Is the test statistically significant?

```python
import pandas as pd 
import scipy.stats as stats 

df = pd.read_csv("./Americandata.csv") 
print(df) 

contingency_table = df.set_index("Happy")  # make the Yes/No column the row labels instead of 1 and 0
print("\nContingency table:")
print(contingency_table)  
print("\nTotal responses:", contingency_table.values.sum())  # add every cell

print("\nPercent within each profession:")
print((contingency_table / contingency_table.sum() * 100).round(1))  # divide by each column total: % yes and % no per profession

chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table, correction=False)  # returns statistic, p-value, degrees of freedom, expected counts

expected_df = pd.DataFrame(expected, index=contingency_table.index, columns=contingency_table.columns)  # put expected counts in a labeled table
print("\nExpected counts:")
print(expected_df.round(2))  # counts we would see if profession and happiness were unrelated

if (expected >= 5).all():  # chi-square is only trustworthy if every expected count is at least 5
    print("\nAssumption Okay; all counts greater than 5.")
else:
    print("\nWARNING! some expected counts may be less than 5.")

print(f"\nChi-square = {chi2:.2f}")  # total distance between observed and expected counts
print(f"Degrees of freedom = {dof}")  # (rows - 1) x (columns - 1) = 1 x 7
print(f"p-value = {p_value:.2e}")  # chance of a result this extreme if profession and happiness were unrelated

alpha = 0.05 
critical_value = stats.chi2.ppf(1 - alpha, dof)  
print(f"\nCritical value (alpha = {alpha} and degree of freedom = {dof}) is: {critical_value:.2f}")
if chi2 > critical_value:  # our statistic is beyond the cutoff, so reject independence
    print(f"Since {chi2:.2f} > {critical_value:.2f}, profession and happiness ARE related")
else:
    print(f"Since {chi2:.2f} <= {critical_value:.2f}, they ARE NOT related")

if p_value < alpha:  # a p-value below alpha gives the same conclusion
    print(f"p < {alpha}: profession and happiness ARE related")
else:
    print(f"p >= {alpha}: no evidence they are related")

contributions = (contingency_table - expected_df) ** 2 / expected_df
print("\nContribution of each cell to chi-square:")
print(contributions.round(1))  # the biggest numbers show which professions drive the relationship
```
